```python
# Lab 03: Transform Textual and Image Data into Machine Understandable Format
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Sample text data (representing text transformation tasks)
documents = [
    "Artificial Intelligence and Machine Learning are amazing.",
    "Data science and web scraping are essential parts of AI.",
    "Transforming text and images into numerical vectors for models."
]

print("--- 1. Text Transformation using CountVectorizer ---")
count_vec = CountVectorizer()
X_counts = count_vec.fit_transform(documents)
print("Vocabulary:", count_vec.get_feature_names_out())
print("Count Matrix Shape:", X_counts.shape)

print("\n--- 2. Text Transformation using TF-IDF Vectorizer ---")
tfidf_vec = TfidfVectorizer()
X_tfidf = tfidf_vec.fit_transform(documents)
print("TF-IDF Matrix Shape:", X_tfidf.shape)
print("TF-IDF Array:\n", X_tfidf.toarray())
```
