```python
import numpy as np

print("--- CNN Pipeline Demonstration ---\n")

# 1. Input Image (5x5 grayscale pixel grid from slides)[cite: 15]
image = np.array([
    [3, 1, 0, 2, 1],
    [2, 0, 1, 3, 0],
    [4, 1, 0, 1, 2],
    [1, 3, 2, 0, 1],
    [0, 2, 1, 4, 2]
])
print("1. Input Image Shape:", image.shape)

# 2. Convolution Layer using a 3x3 Kernel[cite: 15]
kernel = np.array([
    [1,  0, -1],
    [1,  0, -1],
    [1,  0, -1]
])

def convolve2d(img, k):
    h, w = img.shape
    kh, kw = k.shape
    out_h = h - kh + 1
    out_w = w - kw + 1
    output = np.zeros((out_h, out_w))
    for i in range(out_h):
        for j in range(out_w):
            output[i, j] = np.sum(img[i:i+kh, j:j+kw] * k)
    return output

feature_map = convolve2d(image, kernel)
print("\n2. Convolution Feature Map (3x3):")
print(feature_map)

# 3. ReLU Activation: f(x) = max(0, x)[cite: 15]
relu_output = np.maximum(0, feature_map)
print("\n3. ReLU Output (Negative numbers converted to 0):")
print(relu_output)

# 4. Max Pooling (2x2 windows extracting max values)[cite: 15]
# Taking 2x2 sub-regions from the 3x3 feature map
p1 = np.max(relu_output[0:2, 0:2])
p2 = np.max(relu_output[0:2, 1:3])
p3 = np.max(relu_output[1:3, 0:2])
p4 = np.max(relu_output[1:3, 1:3])
pooled_output = np.array([[p1, p2], [p3, p4]])

print("\n4. Max Pooling Output (2x2):")
print(pooled_output)

# 5. Flatten Layer (Converting 2D grid into 1D vector)[cite: 15]
flattened_vector = pooled_output.flatten()
print("\n5. Flattened 1D Vector:")
print(flattened_vector)

# 6. Softmax Output (Converting logits into percentage probabilities)[cite: 15]
def softmax(logits):
    exp_vals = np.exp(logits - np.max(logits))
    return exp_vals / np.sum(exp_vals)

# Example Logits for classes (Cat, Dog, Bird) from slides[cite: 15]
logits = np.array([3.3, 1.5, 0.6])
probabilities = softmax(logits) * 100

print("\n6. Softmax Prediction Probabilities:")
print(f"Cat: {probabilities[0]:.2f}%")
print(f"Dog: {probabilities[1]:.2f}%")
print(f"Bird: {probabilities[2]:.2f}%")
print("Prediction: Cat wins with the highest probability![cite: 15]")
```
